# **<font color=000000>Logical Operators**
Logical operators are used to combine one or more conditional statements and return either Ture or False based on the evaluation of the conditions.


**Name:** <font color=0065a6>AND

**Symbol:** and

**Purpose:** Returns True if both the operands are true

```python
num1=12
num2=13
print(num1 < num2 and num1 > 10)
```

**Name:** <font color=0065a6>OR

**Symbol:** or

**Purpose:** Returns True if either the opearnds are true

```python
num1=16
num2=13
print(num1 < num2 and num1 > 10)
```

**Name:** <font color=0065a6>NOT

**Symbol:** not

**Purpose:** Returns True if the operand is False and vice versa. It reverses the result

```python
num1=12
num2=13
print(not(num1 > 10))
```

# **<font color=000000>Assignment Operator**
The assignment operator is used to assign a value to a variable or a constant. The '=' sign is used as an assignment operator in Python
