# **<font color=000000>Logical Operators**
Logical operators are used to combine one or more conditional statements and return either Ture or False based on the evaluation of the conditions.


**Name:** <font color=0065a6>AND

**Symbol:** and

**Purpose:** Returns True if both the operands are true

```python
num1=12
num2=13
print(num1 < num2 and num1 > 10)
```

**Name:** <font color=0065a6>OR

**Symbol:** or

**Purpose:** Returns True if either the opearnds are true

```python
num1=16
num2=13
print(num1 < num2 and num1 > 10)
```

**Name:** <font color=0065a6>NOT

**Symbol:** not

**Purpose:** Returns True if the operand is False and vice versa. It reverses the result

```python
num1=12
num2=13
print(not(num1 > 10))
```

## **<font color=000000>Augmented Assignment Operator**
The augmented assignment operator takes the value of the operand on the right-hand side and assigns a new value to the operand on the left-hand side. For example, if sum = 0, then sum += 10 is the same as sum = sum + 10.  
These operators with examples are shown below.

**Symbol:** +=

**Purpose:** Adds first and assign a new value

```python
a=12
a+=4
print(a)
```

**Symbol:** -=

**Purpose:** Subtracts first and assign a new value

```python
a=12
a-=4
print(a)
```

**Symbol:** +=

**Purpose:** Multiples first and assign a new value

```python
a=12
a*=4
print(a)
```

**Symbol:** /=

**Purpose:** Divides first and assign a new value

```python
a=12
a/=4
print(a)
```

**Symbol:** //=

**Purpose:** Floor division first and assign a new value

```python
a=12
a//=4
print(a)
```

**Symbol:** %=

**Purpose:** Remainder of a number first and assign a new value

```python
a=12
a%=4
print(a)
```

**Symbol:** **=

**Purpose:** Exponential of a number first and assign a new value

```python
a=12
a**=4
print(a)
```

# **<font color=a66500>Operator Precedence**
An expression is made up of values, variables, operators, and functions. For example, 22 / 7.5 * 5 is an expression. To evaluate an expression with multiple operators, we follow the order of precedence. In Python, this order can be altered by writing an expression with parentheses.  
The order of precedence in the decreasing order is listed below.



|Order of precedence|Operators|
|-------------------|---------|
|1|()|
|2|**|
|3|*/%//|
|4|+-|
|5|<= < > >=|
|6|== !=|
|7|= %= /= //= -= += *= **=|
|8|not,and,or|

 In the above table, as you have seen that there are a few operators with same precedence, then the order of evaluation is from left to right except for the exponential (**) which is always from right to left.
