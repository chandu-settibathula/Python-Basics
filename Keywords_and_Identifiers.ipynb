# **<font color=a66500>Identifiers**
Identifiers are user-defined names of variables, lists, classes, functions, dictionaries, tuples, etc., which are used to identify an element in Python.  
Naming the identifiers follows rules as given below.
*  An identifier can have letters in uppercase or lowercase, numbers, and underscore.
*  It can start with an alphabet or an underscore.
*  No special characters like $, preiod(.), space, etc. are allowed.
*  It should not be a Python keyword.
*  They are case-sensitive. Uppercase and lowercase letters are different.
*  Identifiers can be of any length.    

Some of the examples of valid identifiers are Num1, FIRST_NAME, value, and Scores_first_term.  
Some of the examples of invalid identifiers are 1st_subject, my name, #mobile, city&code, and True.
