# **<font color=a66500>Operators**
Operators are used to perform operations on variables and values. Let us study in detail the commonly used operators in Python.
## **Arithmetic Operators**
Arithmetic operators are used to perform mathematical operations on variables and values. Arithmetic operators used in Python are


**Name:** <font color=0065a6>Addition

**Symbol:** +

**Purpose:** Adds two numbers (With strings it concentrates two values)

```python
a=12
b=32
print(a+b)
```

```python
a="Good "
b="Morning"
print(a+b)
```

**Name:** <font color=0065a6>Subtraction

**Symbol:** -

**Purpose:** Subtracts second value from first value

```python
a=12.8
b=2
print(a-b)
```

**Name:** <font color=0065a6>Multiplication

**Symbol:** *

**Purpose:** Multiplies two values (strings * int repeats the string that many times)

```python
a=3
b=4
print(a*b)
```

```python
a="Python "
print(a*4)
```

**Name:** <font color=0065a6>Division

**Symbol:** /

**Purpose:** Divides two numbers and returns decimal value

```python
a=9
b=3
print(a/b)
```

**Name:** <font color=0065a6>Floor division

**Symbol:** //

**Purpose:** Divides two numbers and returns integer value

```python
a=9
b=3
print(a//b)
```

**Name:** <font color=0065a6>Remainder

**Symbol:** %

**Purpose:** Returns the remainder of the division

```python
a=17
b=4
print(a%b)
```

**Name:** <font color=0065a6>Exponential

**Symbol:** **

**Purpose:** First number raised to the power second number

```python
a=2
b=5
print(a**b)
```

## **Relational Operators**
Relational or comparison operators evaluate the values on both sides of the operator and return a Boolean result: True if the condition is met and False otherwise. Relational operators used in Python are

**Name:** <font color=0065a6>Equal to

**Symbol:** ==

**Purpose:** Returns True if both values are same

```python
a=12
b=11
print(a==b)
```

**Name:** <font color=0065a6>Not equal to

**Symbol:** !=

**Purpose:** Returns True if both values are different

```python
a=12
b=11
print(a!=b)
```

**Name:** <font color=0065a6>Greater than

**Symbol:** >

**Purpose:** Returns True if first value is greater than the second value

```python
a=12
b=11
print(a>b)
```

**Name:** <font color=0065a6>Less than

**Symbol:** <

**Purpose:** Returns True if first value is less than the second value

```python
a=12
b=11
print(a<b)
```

**Name:** <font color=0065a6>Greater than or equal to

**Symbol:** >=

**Purpose:** Returns True if first value is greater than or equals than the second value

```python
a=8
b=9
print(a>=b)
```

**Name:** <font color=0065a6>Less than or equal to

**Symbol:** <=

**Purpose:** Returns True if first value is less than or equals than the second value

```python
a=11
b=11
print(a<=b)
```
