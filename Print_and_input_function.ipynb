# **<font color=a66500>The print() Function**
The print() function is used to display an output on the screen. It converts the expressions into a string before writing to the screen. Syntax of print() function is  
**Syntax:** print(object(s),sep=separator,end=end)  

*  where *object* can be more or one seperated by comma and it can be a variable.Object is converted to string before printing
*  where *sep* seperator between the items to print,Default sep is space
*  where *end* allows to specify what to print at the end of the output.default is newline character ('\n') means each print statement ends with a new line  

Let us see some examples

```python
print("Good")
```

```python
print("Good","Morning")
```

```python
print("hi "+"how are you")#(This is concatenation of string using "+" operator)
```

```python
print("trends","following",sep="$")
```

```python
print("hello","friends",end="!")
```

```python
print("hello")
```

```python
print()
```

```python
print("Pythons")
```

```python
print("hello","friends","welcome",sep="%",end="@")
```

```python
a=5
b=10
print("sum of two numbers is",str(a+b))
```

```python
age=19
print("I am",age,"years old")
```

# **<font color=a66500>The Input() Function**
This is used to take input from the user.
When this statement executes, the flow of program stops and waits with the cursor blinking for the user to enter a vlaue
The input given by user is always a string value in python.
Syntax of input() function is  
**Syntax:** input(prompt)  
Let us see some examples

```python
name=input("enter your name ")
print(name)
```

```python
N1=input("Enter first number: ")
N2=input("Enter second number: ")
sum=N1 + N2
print("sum = ",sum)
```

```python
N1=int(input("Enter first number:"))
N2=int(input("Enter second number:"))
sum=N1+N2
print("The sum of ",N1,"and ",N2,"is ",sum)
```
